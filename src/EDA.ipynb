```python
# 1. Створюємо генератор випадкових значень
rng = np.random.default_rng(42)

# 2. Визначаємо часові межі 2025 року
start_time = pd.Timestamp("2025-01-01 00:00:00")
end_time = pd.Timestamp("2026-01-01 00:00:00")

# 3. Генеруємо випадкові дати та час для кожного запису
seconds_in_year = int((end_time - start_time).total_seconds())

df_StressLevelDataset["timestamp"] = (
    start_time
    + pd.to_timedelta(
        rng.integers(
            0, seconds_in_year,
            size=len(df_StressLevelDataset)
        ),
        unit="s"
    )
)

# 4. Перевіряємо результат
print(df_StressLevelDataset[["timestamp", "stress_level"]].head())

print("Найдавніша дата:", df_StressLevelDataset["timestamp"].min())
print("Найновіша дата:", df_StressLevelDataset["timestamp"].max())

# 5. Зберігаємо датасет із новою колонкою
df_StressLevelDataset.to_csv(
    "StressLevelDataset_clean.csv",
    index=False
)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# Підключення до SQL Server через SQLAlchemy
connection_string = quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=Oleg;"
    "DATABASE=StudentStressMonitoring;"
    "Trusted_Connection=yes;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={connection_string}"
)

# Збереження очищених датасетів у нові таблиці SQL Server
df_Stress_Dataset.to_sql(
    "Stress_DatasetClean",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False
)

df_StressLevelDataset.to_sql(
    "StressLevelDatasetClean",
    con=engine,
    schema="dbo",
    if_exists="replace",
    index=False
)

print("Очищені датасети збережено в SQL Server!")
```
